```python
import pandas as pds
from matplotlib import pyplot as plt
import numpy as np
```

```python
df = pds.read_csv("../data/Life_Expectancy_at_Birth.csv", index_col = "Country")



df = df.drop(["ISO3", "Hemisphere", "Human Development Groups", "UNDP Developing Regions",
             "HDI Rank (2021)"], axis = 1)

x_values = range(1990,2022)

all_continents = df["Continent"].value_counts()
for i in range(0,all_continents.count()): 
    one_continent_df = df[df["Continent"] ==  all_continents.index[i]]
    y_values = one_continent_df.mean(numeric_only = True).values
    plt.plot(x_values, y_values, label = all_continents.index[i])
    

plt.title("World Life Expectancy At Birth increases \n from 1990 - 2021")
plt.xlabel("Year")
plt.ylabel("Life Expectancy At Birth")
plt.legend()
plt.grid(axis = "y")
```
