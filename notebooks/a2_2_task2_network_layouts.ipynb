```python
import csv
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.Graph()
edges = csv.reader(open('../data/edges.csv'))
next(edges)
for row in edges:
    G.add_edge(row[0],row[1])
```

```python
pos = nx.circular_layout(G)
nx.draw_networkx(G, pos=pos,
                 font_weight = "bold",
                 font_color = "white")
plt.title('Radial layout')
```

```python
init_pos = nx.random_layout(G, seed=1)
pos = nx.kamada_kawai_layout(G)
nx.draw_networkx(G, pos=init_pos, 
                 font_weight = "bold",
                 font_color = "white" )
plt.title('Kamada-Kawai layout (seed = 1)')
```

```python
pos = nx.fruchterman_reingold_layout(G,seed = 1)
nx.draw_networkx(G, pos=pos,
                 font_weight = "bold",
                 font_color = "white" )
plt.title('Fruchterman-Reingold layout (seed = 1)')
```
